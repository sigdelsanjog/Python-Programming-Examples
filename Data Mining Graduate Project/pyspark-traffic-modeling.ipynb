```python
trafficPath = 'file:///home/travelingnepal/Downloads/Dodgers.data'
gamesPath = 'file:///home/travelingnepal/Downloads/Dodgers.events'
```

```python
traffic = sc.textFile(trafficPath)
traffic.take(10)
```

```python
games= sc.textFile(gamesPath)
games.take(10)
```

```python
from datetime import datetime
import csv
from io import StringIO


def parseTraffic(row):
    DATE_FMT = "%m/%d/%Y %H:%M"
    row = row.split(",")
    row[0] = datetime.strptime(row[0], DATE_FMT)
    row[1] = int(row[1])
    return (row[0],row[1])
```

```python
trafficParsed = traffic.map(parseTraffic)
```

```python
# The result is a pair RDDs where each entry is a tuple
trafficParsed.take(10)
```

```python
dailyTrend = trafficParsed.map(lambda x:(x[0].date(),x[1]))\
            .reduceByKey(lambda x,y:x+y)
```

```python
dailyTrend.take(10)
```

```python
dailyTrend.sortBy(lambda x:-x[1]).take(10)
```

```python
def parseGames(row):
    DATE_FMT = "%m/%d/%y"
    row = row.split(",")
    row[0] = datetime.strptime(row[0], DATE_FMT).date()
    return (row[0],row[4])


gamesParsed = games.map(parseGames)
gamesParsed.take(10)
```

```python
dailyTrendCombined = dailyTrend.leftOuterJoin(gamesParsed)
```

```python
dailyTrendCombined.take(10)
```

```python
dailyTrendCombined = dailyTrend.join(gamesParsed)
```

```python
dailyTrendCombined.take(10)
```

```python
dailyTrendCombined = dailyTrend.rightOuterJoin(gamesParsed)
```

```python
dailyTrendCombined.take(10)
```

```python
def checkGameDay(row):
        if row[1][1] ==None:
            return (row[0],row[1][1],"Regular Day", row[1][0])
        else:
            return (row[0],row[1][1], "Game Day", row[1][0])
dailyTrendbyGames = dailyTrendCombined.map(checkGameDay)
```

```python
dailyTrendbyGames.take(10)
```

```python
dailyTrendbyGames.sortBy(lambda x:-x[3]).take(10)
```
